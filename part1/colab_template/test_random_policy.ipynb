```python
import gymnasium as gym
```

```python
env = gym.make('Hopper-v5', render_mode='rgb_array')

print('State space:', env.unwrapped.observation_space)  # state-space
print('Action space:', env.action_space)  # action-space
```

```python
n_episodes = 5

for episode in range(n_episodes):
  done = False
  observation = env.reset()	 # Reset environment to initial state

  while not done:  # Until the episode is over

    action = env.action_space.sample()	# Sample random action

    observation, reward, terminated, truncated, _ = env.step(action)	# Step the simulator to the next
    done = terminated or truncated
```

```python
import gymnasium as gym
import imageio
import os

# --- PASSO 1: Sblocchiamo il file se esiste già ---
# Prova a chiudere l'ambiente se era rimasto aperto
if 'env' in locals():
    env.close()

fname = 'hopper_random.gif'
# Prova a cancellare il file vecchio per assicurarci che non sia bloccato
try:
    if os.path.exists(fname):
        os.remove(fname)
        print(f"File vecchio '{fname}' rimosso con successo.")
except OSError as e:
    print(f"ATTENZIONE: Non riesco a cancellare il file '{fname}'.")
    print(f"Errore di Windows: {e}")
    print("--> Per favore, vai nel menu 'Kernel' -> 'Restart', poi riprova.")
    raise # Ferma l'esecuzione se il file è bloccato

# --- PASSO 2: Preparazione Ambiente ---
print("\nInizializzazione ambiente...")
env = gym.make('Hopper-v5', render_mode='rgb_array')
observation, info = env.reset()

frames = []
max_steps = 300 # Aumentiamo gli step massimi

# --- PASSO 3: Loop di simulazione con diagnostica ---
print("Inizio simulazione con azioni casuali...")
for step in range(max_steps):
    # Cattura il frame attuale
    frame = env.render()
    if frame is not None:
        frames.append(frame)
    else:
        print(f"Errore: render() ha restituito None al passo {step}")
        break

    # Azione CASUALE
    action = env.action_space.sample()
    observation, reward, terminated, truncated, info = env.step(action)

    # Diagnostica: se finisce subito, stampiamo il perché
    if terminated or truncated:
        print(f"-> Simulazione TERMINATA al passo {step}.")
        if terminated: print("   Motivo: Il robot è caduto (Terminated).")
        if truncated: print("   Motivo: Tempo scaduto (Truncated).")
        
        # --- TRUCCO PER VEDERE IL MOVIMENTO ---
        # Se cade troppo presto (es. meno di 10 frame), 
        # non fermiamo il loop, continuiamo a fare render() 
        # per un po' per mostrare la caduta.
        if len(frames) < 20:
             print("   (Raccogo altri frame della caduta...)")
             continue 
        else:
            break # Esce dal loop se abbiamo abbastanza frame

env.close()

# --- PASSO 4: Salvataggio con verifica ---
num_frames = len(frames)
print(f"\nTotale fotogrammi raccolti: {num_frames}")

if num_frames > 1:
    print(f"Salvataggio della GIF '{fname}' (loop infinito)...")
    # loop=0 assicura che la GIF si ripeta continuamente
    imageio.mimsave(fname, frames, fps=30, loop=0)
    print("Salvataggio completato! Controlla il file.")
elif num_frames == 1:
    print("Errore: Ho raccolto solo 1 fotogramma. La simulazione è finita troppo presto.")
    print("Riprova a eseguire la cella, l'azione casuale potrebbe essere migliore la prossima volta.")
else:
    print("Errore critico: Nessun fotogramma raccolto.")
```
